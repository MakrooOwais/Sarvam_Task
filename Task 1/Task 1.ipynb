````python
import whisper
model = whisper.load_model("medium.en", device = "cuda")
audio = whisper.load_audio('audio.mp3')
result = model.transcribe(audio, word_timestamps = True, verbose = False)
````

````python
result = result['segments']
````

````python
import json
with open("/content/transcript.json", 'w') as f:
  json.dump(result, f, indent = 4)
````

````python
def break_transcript(dict):
    chs = ['.', '?', ',']
    idxs = [i for i, letter in enumerate(dict['text']) if letter in chs]
    i = min(idxs, key=lambda x:abs(x-(len(dict['text'])//2)))

    first = {'start': dict['start']}
    second = {'end': dict['end']}
    first['text'] = ''
    for j in range(len(dict['words'])):
        if len(first['text']) < i:
            first['text'] += dict['words'][j]['word']
            first['end'] = dict['words'][j]['end']
        else:
            end_idx = j
            break
    first['words'] = dict['words'][0:end_idx]
    second['words'] = dict['words'][end_idx:]

    second['text'] = ''
    second['start'] = dict['words'][end_idx]['start']
    for x in range(end_idx, len(dict['words'])):
        second['text'] += dict['words'][x]['word']

    return first, second
````

````python
final_results = list()
for i in result:
    if i['end'] - i['start'] < 15:
        final_results.append({
        "chunk_id": len(final_results) + 1,
        "chunk_length": i['end'] - i['start'],
        "text": i['text'],
        "start_time": i['start'],
        "end_time": i['end'],
    },)
    if i['end'] - i['start'] >= 15:
        first, second = break_transcript(i)
        if first['end'] - first['start'] >= 15:
            firsts = list(break_transcript(first))
        else:
            firsts = [first]

        if second['end'] - second['start'] >= 15:
            seconds = list(break_transcript(second))
        else:
            seconds = [second]

        for first in firsts:
            final_results.append({
            "chunk_id": len(final_results) + 1,
            "chunk_length": first['end'] - first['start'],
            "text": first['text'],
            "start_time": first['start'],
            "end_time": first['end'],
            },)
        for second in seconds:
            final_results.append({
            "id": len(final_results) + 1,
            "length": second['end'] - second['start'],
            "text": second['text'],
            "start_time": second['start'],
            "end_time": second['end'],
            },)
````
